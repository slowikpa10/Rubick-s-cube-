```python
import random
ls = ['L', "L'",'R',"R'", 'U', "U'", 'D', "D'", 'F', "F'", 'B',"B'",'L2','U2','R2','D2','B2','F2']
```

```python
X= ['B', "B'", 'B2']
Y= ['F', "F'", 'F2']
Z= ['D', "D'", 'D2']
V= ['U', "U'", 'U2']
W= ['L', "L'", 'L2']
S= ['R', "R'", 'R2']


wynik =[]
ostatni = None #nie mamy jeszcze ostatniego wyniku bo nie losowalam jeszcze
for _ in range(19): #_ nie interesuje zmienna
  mozliwe =[]
  for x in ls:
    if x==ostatni:# czy x jest taki sam jak poprzedni
      continue #pomijamy i losujemy cos innego jesli sa takie same 
    
    if (ostatni in X and x in X) or (ostatni in Y and x in Y) or (ostatni in Z and x in Z) or (ostatni in V and x in V) or (ostatni in W and x in W) or (ostatni in S and x in S):
      continue
    
    mozliwe.append(x)

  losowanie = random.choice(mozliwe)

  wynik.append(losowanie)
  ostatni = losowanie

print(*wynik, sep=" , ") #* - unpacking iterables 
```
